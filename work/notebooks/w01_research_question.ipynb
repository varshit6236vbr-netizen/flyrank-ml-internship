Why I chose this lane:
SEO and editorial teams manage hundreds or thousands of published pages across client websites. ## 2. The question: decision, action, cost of a wrong call

*What decision does your work improve? Who acts on it? What does a wrong recommendation cost?*

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
```

Research Question: "Which high-performing client pages are exhibiting signs of content decay and should be queued for a content refresh this month?"

Unit of Analysis: A single published page/URL (page_id) over a 30-day evaluation window.

Decision: Deciding whether an editorial team should allocate writer hours and budget to rewrite/refresh a specific page or leave it untouched.

Action: Automatically flagging the page and adding it to the Monthly Content Refresh Queue assigned to content strategists.

Cost of a Wrong Call:

False Positive (Flagging a page that didn't need a refresh): Wasted editorial budget (~$200–$500 per unnecessary rewrite), wasted writer hours, and the opportunity cost of neglecting pages that actually needed attention.

False Negative (Missing a declining page): Compounding loss of organic search visibility, leading to fewer organic leads/conversions and forcing the business to rely on costly paid ads to replace lost organic traffic.

Why Data / ML Helps: Simple rule-based logic (e.g., "if traffic dropped 10%") fails because it cannot separate regular seasonal fluctuations from true content decay. Machine learning can learn complex, multi-variable patterns across rankings, impression trends, CTR, and search intent shifts to accurately predict decay before severe organic traffic drops occur.
